### Clase ConexionDB

````python
%%writefile /content/book_manager/src/book_manager/database/connection.py
"""Conexión a la base de datos mediante SQLAlchemy."""

import os
from pathlib import Path
from typing import Optional
from urllib.parse import quote_plus

from dotenv import load_dotenv
from sqlalchemy import create_engine, text
from sqlalchemy.engine import Engine
from sqlalchemy.exc import SQLAlchemyError
from sqlalchemy.orm import Session

RAIZ_PROYECTO = Path(__file__).resolve().parents[3]
RUTA_ENV = RAIZ_PROYECTO / ".env"
VARIABLE_URL = "DATABASE_URL"
VARIABLE_PASSWORD = "DATABASE_PASSWORD"
MARCADOR_PASSWORD = "[YOUR-PASSWORD]"


class ConexionDB:
    """Maneja la conexión a la base de datos usando SQLAlchemy."""

    def __init__(self, url: Optional[str] = None, echo: bool = False) -> None:
        """Crea el engine de SQLAlchemy.

        Args:
            url: URL de conexión. Si no se indica, se lee DATABASE_URL
                del archivo .env.
            echo: En True muestra por pantalla las sentencias SQL ejecutadas.

        Raises:
            ValueError: Si falta la URL o la contraseña de la base.
        """
        url_completa = self._resolver_url(url)
        self._engine: Engine = create_engine(url_completa, echo=echo)

    @property
    def url(self) -> str:
        """URL de conexión en uso, con la contraseña oculta."""
        return self._engine.url.render_as_string(hide_password=True)

    @property
    def engine(self) -> Engine:
        """Engine de SQLAlchemy (solo lectura)."""
        return self._engine

    def obtener_sesion(self) -> Session:
        """Devuelve una sesión nueva para operar con la base de datos."""
        return Session(self._engine)

    def probar_conexion(self) -> bool:
        """Verifica que se pueda conectar a la base de datos.

        Returns:
            True si la conexión funciona, False en caso contrario.
        """
        try:
            with self._engine.connect() as conexion:
                conexion.execute(text("SELECT 1"))
            return True
        except SQLAlchemyError:
            return False

    def cerrar(self) -> None:
        """Libera las conexiones abiertas por el engine."""
        self._engine.dispose()

    @staticmethod
    def _resolver_url(url: Optional[str]) -> str:
        """Obtiene y valida la URL de conexión.

        Si la URL trae el marcador [YOUR-PASSWORD], lo reemplaza por la
        contraseña guardada en la variable de entorno DATABASE_PASSWORD.
        """
        load_dotenv(RUTA_ENV)
        if url is None:
            url = os.getenv(VARIABLE_URL)

        if not url or not url.strip():
            raise ValueError(
                "No se encontró la URL de la base de datos. "
                f"Definir {VARIABLE_URL} en el archivo .env."
            )
        url = url.strip()

        if MARCADOR_PASSWORD in url:
            password = os.getenv(VARIABLE_PASSWORD)
            if not password:
                raise ValueError(
                    "Falta la contraseña de la base de datos. "
                    f"Definir la variable de entorno {VARIABLE_PASSWORD}."
                )
            url = url.replace(MARCADOR_PASSWORD, quote_plus(password))
        return url
````

Crear un context manager para manejar las transacciones a la base de datos.

### Archivo connection.py: context manager Transaccion

````python
%%writefile /content/book_manager/src/book_manager/database/connection.py
"""Conexión y transacciones a la base de datos mediante SQLAlchemy."""

import os
from pathlib import Path
from types import TracebackType
from typing import Optional, Type
from urllib.parse import quote_plus

from dotenv import load_dotenv
from sqlalchemy import create_engine, text
from sqlalchemy.engine import Engine
from sqlalchemy.exc import SQLAlchemyError
from sqlalchemy.orm import Session

RAIZ_PROYECTO = Path(__file__).resolve().parents[3]
RUTA_ENV = RAIZ_PROYECTO / ".env"
VARIABLE_URL = "DATABASE_URL"
VARIABLE_PASSWORD = "DATABASE_PASSWORD"
MARCADOR_PASSWORD = "[YOUR-PASSWORD]"


class ConexionDB:
    """Maneja la conexión a la base de datos usando SQLAlchemy."""

    def __init__(self, url: Optional[str] = None, echo: bool = False) -> None:
        """Crea el engine de SQLAlchemy.

        Args:
            url: URL de conexión. Si no se indica, se lee DATABASE_URL
                del archivo .env.
            echo: En True muestra por pantalla las sentencias SQL ejecutadas.

        Raises:
            ValueError: Si falta la URL o la contraseña de la base.
        """
        url_completa = self._resolver_url(url)
        self._engine: Engine = create_engine(url_completa, echo=echo)

    @property
    def url(self) -> str:
        """URL de conexión en uso, con la contraseña oculta."""
        return self._engine.url.render_as_string(hide_password=True)

    @property
    def engine(self) -> Engine:
        """Engine de SQLAlchemy (solo lectura)."""
        return self._engine

    def obtener_sesion(self) -> Session:
        """Devuelve una sesión nueva para operar con la base de datos.

        Con expire_on_commit=False los objetos leídos se pueden seguir
        usando después de confirmar y cerrar la sesión.
        """
        return Session(self._engine, expire_on_commit=False)

    def transaccion(self) -> "Transaccion":
        """Devuelve un context manager para operar dentro de una transacción.

        Ejemplo:
            with conexion.transaccion() as sesion:
                sesion.add(objeto)
        """
        return Transaccion(self)

    def probar_conexion(self) -> bool:
        """Verifica que se pueda conectar a la base de datos.

        Returns:
            True si la conexión funciona, False en caso contrario.
        """
        try:
            with self._engine.connect() as conexion:
                conexion.execute(text("SELECT 1"))
            return True
        except SQLAlchemyError:
            return False

    def cerrar(self) -> None:
        """Libera las conexiones abiertas por el engine."""
        self._engine.dispose()

    @staticmethod
    def _resolver_url(url: Optional[str]) -> str:
        """Obtiene y valida la URL de conexión.

        Si la URL trae el marcador [YOUR-PASSWORD], lo reemplaza por la
        contraseña guardada en la variable de entorno DATABASE_PASSWORD.
        """
        load_dotenv(RUTA_ENV)
        if url is None:
            url = os.getenv(VARIABLE_URL)

        if not url or not url.strip():
            raise ValueError(
                "No se encontró la URL de la base de datos. "
                f"Definir {VARIABLE_URL} en el archivo .env."
            )
        url = url.strip()

        if MARCADOR_PASSWORD in url:
            password = os.getenv(VARIABLE_PASSWORD)
            if not password:
                raise ValueError(
                    "Falta la contraseña de la base de datos. "
                    f"Definir la variable de entorno {VARIABLE_PASSWORD}."
                )
            url = url.replace(MARCADOR_PASSWORD, quote_plus(password))
        return url


class Transaccion:
    """Context manager para manejar las transacciones a la base de datos.

    Al entrar al bloque with abre una sesión. Al salir confirma los cambios
    (commit) si no hubo errores o los deshace (rollback) si los hubo. En
    ambos casos cierra la sesión.
    """

    def __init__(self, conexion: ConexionDB) -> None:
        """Guarda la conexión con la que se abrirá la sesión.

        Raises:
            TypeError: Si no se recibe una instancia de ConexionDB.
        """
        if not isinstance(conexion, ConexionDB):
            raise TypeError("Se esperaba una instancia de ConexionDB.")
        self._conexion: ConexionDB = conexion
        self._sesion: Optional[Session] = None

    def __enter__(self) -> Session:
        """Abre la sesión y la entrega al bloque with."""
        self._sesion = self._conexion.obtener_sesion()
        return self._sesion

    def __exit__(
        self,
        exc_type: Optional[Type[BaseException]],
        exc_value: Optional[BaseException],
        traceback: Optional[TracebackType],
    ) -> bool:
        """Confirma o deshace los cambios y cierra la sesión.

        Returns:
            False, para que si hubo una excepción se siga propagando.
        """
        try:
            if exc_type is None:
                self._sesion.commit()
            else:
                self._sesion.rollback()
        finally:
            self._sesion.close()
            self._sesion = None
        return False
````

````python
#@title Prueba del context manager de transacciones
import importlib
from sqlalchemy import text
import book_manager.database.connection as modulo_conexion

importlib.reload(modulo_conexion)  # toma la última versión del archivo
from book_manager.database.connection import ConexionDB

conexion = ConexionDB()

with conexion.transaccion() as sesion:
  sesion.execute(text("DROP TABLE IF EXISTS prueba_transaccion"))
  sesion.execute(text("CREATE TABLE prueba_transaccion (id INTEGER)"))

# 1) Sin errores: los cambios se confirman (commit)
with conexion.transaccion() as sesion:
  sesion.execute(text("INSERT INTO prueba_transaccion VALUES (1)"))

# 2) Con error: los cambios se deshacen (rollback)
try:
  with conexion.transaccion() as sesion:
    sesion.execute(text("INSERT INTO prueba_transaccion VALUES (2)"))
    raise RuntimeError("error simulado")
except RuntimeError as error:
  print("Se produjo un", error, "-> se deshizo la transacción")

with conexion.transaccion() as sesion:
  filas = sesion.execute(text("SELECT id FROM prueba_transaccion")).all()
  print("Filas guardadas:", filas)
  sesion.execute(text("DROP TABLE prueba_transaccion"))

conexion.cerrar()
````

````python
#@title Definición de las tablas
%%writefile /content/book_manager/src/book_manager/models/models.py

"""Modelos ORM y esquema relacional de Book Manager."""

from __future__ import annotations

from datetime import date
from decimal import Decimal

from sqlalchemy import (
    CheckConstraint,
    Date,
    ForeignKey,
    Integer,
    Numeric,
    String,
    UniqueConstraint,
)
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column, relationship


class Base(DeclarativeBase):
    """Base declarativa compartida por los modelos ORM."""


class Genero(Base):
    """Género literario al que pertenece un libro."""

    __tablename__ = "generos"
    __table_args__ = (UniqueConstraint("nombre", name="uq_generos_nombre"),)

    id_genero: Mapped[int] = mapped_column(Integer, primary_key=True)
    nombre: Mapped[str] = mapped_column(String(100), nullable=False)

    libros: Mapped[list[Libro]] = relationship(back_populates="genero")


class Editorial(Base):
    """Editorial que publica los libros."""

    __tablename__ = "editoriales"
    __table_args__ = (
        UniqueConstraint("nombre", name="uq_editoriales_nombre"),
    )

    id_editorial: Mapped[int] = mapped_column(Integer, primary_key=True)
    nombre: Mapped[str] = mapped_column(String(150), nullable=False)

    libros: Mapped[list[Libro]] = relationship(back_populates="editorial")


class Moneda(Base):
    """Moneda en la que se expresa un precio."""

    __tablename__ = "monedas"
    __table_args__ = (
        UniqueConstraint("codigo", name="uq_monedas_codigo"),
    )

    id_moneda: Mapped[int] = mapped_column(Integer, primary_key=True)
    codigo: Mapped[str] = mapped_column(String(3), nullable=False)
    descripcion: Mapped[str] = mapped_column(String(100), nullable=False)

    precios: Mapped[list[Precio]] = relationship(back_populates="moneda")


class TipoCotizacion(Base):
    """Tipo de cotización del dólar."""

    __tablename__ = "tipos_cotizacion"
    __table_args__ = (
        UniqueConstraint("nombre", name="uq_tipos_cotizacion_nombre"),
    )

    id_tipo_cotizacion: Mapped[int] = mapped_column(
        Integer, primary_key=True
    )
    nombre: Mapped[str] = mapped_column(String(100), nullable=False)

    cotizaciones: Mapped[list[CotizacionDolar]] = relationship(
        back_populates="tipo_cotizacion"
    )


class Libro(Base):
    """Libro del catálogo."""

    __tablename__ = "libros"
    __table_args__ = (
        UniqueConstraint("isbn", name="uq_libros_isbn"),
    )

    id_libro: Mapped[int] = mapped_column(Integer, primary_key=True)
    isbn: Mapped[str] = mapped_column(String(13), nullable=False)
    titulo: Mapped[str] = mapped_column(String(255), nullable=False)
    autor: Mapped[str] = mapped_column(String(255), nullable=False)
    id_genero: Mapped[int] = mapped_column(
        ForeignKey("generos.id_genero"), nullable=False
    )
    id_editorial: Mapped[int] = mapped_column(
        ForeignKey("editoriales.id_editorial"), nullable=False
    )

    genero: Mapped[Genero] = relationship(back_populates="libros")
    editorial: Mapped[Editorial] = relationship(back_populates="libros")
    precios: Mapped[list[Precio]] = relationship(back_populates="libro")
    stock: Mapped[Stock | None] = relationship(back_populates="libro")


class Precio(Base):
    """Precio de un libro en una moneda."""

    __tablename__ = "precios"
    __table_args__ = (
        UniqueConstraint(
            "id_libro", "id_moneda", name="uq_precios_libro_moneda"
        ),
        CheckConstraint("monto > 0", name="ck_precios_monto_positivo"),
    )

    id_precio: Mapped[int] = mapped_column(Integer, primary_key=True)
    id_libro: Mapped[int] = mapped_column(
        ForeignKey("libros.id_libro"), nullable=False
    )
    monto: Mapped[Decimal] = mapped_column(Numeric(12, 2), nullable=False)
    id_moneda: Mapped[int] = mapped_column(
        ForeignKey("monedas.id_moneda"), nullable=False
    )

    libro: Mapped[Libro] = relationship(back_populates="precios")
    moneda: Mapped[Moneda] = relationship(back_populates="precios")


class Stock(Base):
    """Existencia de un libro, identificada por su clave foránea."""

    __tablename__ = "stock"
    __table_args__ = (
        CheckConstraint("cantidad >= 0", name="ck_stock_cantidad_no_negativa"),
    )

    id_libro: Mapped[int] = mapped_column(
        ForeignKey("libros.id_libro"), primary_key=True
    )
    cantidad: Mapped[int] = mapped_column(Integer, nullable=False)

    libro: Mapped[Libro] = relationship(back_populates="stock")


class CotizacionDolar(Base):
    """Cotización histórica identificada por tipo y fecha."""

    __tablename__ = "cotizaciones_dolar"
    __table_args__ = (
        CheckConstraint("valor > 0", name="ck_cotizaciones_valor_positivo"),
    )

    id_tipo_cotizacion: Mapped[int] = mapped_column(
        ForeignKey("tipos_cotizacion.id_tipo_cotizacion"), primary_key=True
    )
    fecha: Mapped[date] = mapped_column(Date, primary_key=True)
    valor: Mapped[Decimal] = mapped_column(Numeric(12, 4), nullable=False)

    tipo_cotizacion: Mapped[TipoCotizacion] = relationship(
        back_populates="cotizaciones"
    )
````
